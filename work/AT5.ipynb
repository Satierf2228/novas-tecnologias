### Gráfico da váriavel 'Survival'

```python
sns.countplot(x = 'Survived', data = df)

plt.title("Sobreviventes do Titanic")
plt.xlabel("Sobreviveram")
plt.title("Quantidade")

plt.show()
```

### Por quer da escolha do gráfico
-Como a comparação é de pessoas sobreviventes, ou seja, 0 = não sobreviveu e 1 = sobreviveu, o gráfico consegue comparar bem entre a quantidade das duas categorias.

```python
sobreviveu = df['Survived'].value_counts()[1]
nao_sobrev = df['Survived'].value_counts()[0]

print("Sobreviveram: ", sobreviveu)
print("Não sobreviveram: ", nao_sobrev)
```

### Relação ente 'sex' e 'survival'

```python
tabela = pd.crosstab(df['Sex'], df['Survived'])

print(tabela)
```

### 'Sex' x 'Survival'

```python
sns.countplot(x = 'Sex', hue = 'Survived', data = df)

plt.title("Sobrevivência por Sexo")
plt.xlabel("Sexo")
plt.ylabel("Quantidade")

plt.show()
```

Na tabela, demonstra-se que as mulheres tiveram uma taxa de sobrevivência maior do que a dos homens

### Idades dos passageiros

```python
sns.histplot(df['Age'], kde = True)

plt.title("Distribuição das Idades")
plt.xlabel("Idades")
plt.ylabel("Quantidade")

plt.show()
```
